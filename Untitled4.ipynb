```python
# outliers
numeric_cols = [
    "Age","Study_Hours","Attendence","Final_Score","Assignment_Score","Previous_Score"]
```

```python
# STUDY HOURS
Q1=df["Study_Hours"].quantile(0.25)
Q1=df["Study_Hours"].quantile(0.25)
IQR=Q3-Q1
lower= Q1 - 1.5 * IQR
upper= Q3 + 1.5 * IQR
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower)
print("Upper Bound:", upper)
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(x=df["Study_Hours"])
plt.title("boxplot of STUDY hours")
plt.show()
```

```python
# ATTENDENCE
Q1=df["Attendance"].quantile(0.25)
Q3=df["Attendance"].quantile(0.25)
IQR=Q3-Q1
lower= Q1 - 1.5 * IQR
upper= Q3 + 1.5 * IQR
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower)
print("Upper Bound:", upper)
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(x=df["Attendance"])
plt.title("boxplot Attendance")
plt.show()
```

```python
# HISTOGRAM
sns.histplot(df["Final_Score"], kde=True)
plt.title("final score")
plt.xlabel("final score")
plt.show()
```

```python
# HISTOGRAM
sns.histplot(df["Age"], kde=True)
plt.title("Age")
plt.xlabel("AGE")
plt.show()
```

```python
# countplot(gender)

sns.countplot(data=df,x="Gender")
plt.title("gender countplot")
plt.xlabel("gender")
plt.show()
```

```python
df["Average_finalscore"]=df["Final_Score"].mean()
print(df["Average_finalscore"])
```

```python
sns.barplot(data=df,y="Average_finalscore",x="Gender")
plt.title("averagescore vs gender")
plt.xlabel("gender")
plt.show()
```

```python
# scatterplot
```

```python
sns.scatterplot(data=df,x="Study_Hours",y="Final_Score")
plt.title("study hours depends on finalscore")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df.select_dtypes(include="number").corr(),annot=True,vmin=-1, vmax=1,cmap="coolwarm",fmt='2f')
plt.show()
```
